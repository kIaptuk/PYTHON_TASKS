У випадках коли немє значеня/ень для виконання заданого резултату, return відображає стандартний результат який відображається як "None"

````python
dict = {12:6, 20:0, 112:10, 99:0, 11:1.1}

def int_division(a,b):
  if isinstance(a,int) and isinstance(b,int):
    if b == 0:
      print("Не можу виконати ділення на 0")
    else:
      return a/b
  else:
   print("Не працюю з такими типами даних.")

for i in dict:
  print(int_division(i, dict[i]))
````
