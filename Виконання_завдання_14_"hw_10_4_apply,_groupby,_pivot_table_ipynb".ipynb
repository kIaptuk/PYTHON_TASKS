Спочатку визначається співвідношення гендеру покупця до прибутку

```python

df.pivot_table(index = "Gender", columns = 'Product line', values = 'gross income',  aggfunc= 'sum')
```

Де співвідношення в кількості дорівнює:

```python
df.pivot_table(index = "Gender", columns = 'Product line', values = 'gross income',  aggfunc= 'count')
```

Далі шукається співвідношення рейтингу

```python
df.groupby('Branch')['Rating'].mean()
```

І під кінець - середнє значення з типу оплати

```python
df.pivot_table(index = "Customer type", columns = 'Payment', values = 'gross income',  aggfunc= 'sum')
```

```python
df.pivot_table(index = "Customer type", columns = 'Payment', values = 'gross income',  aggfunc= 'count')
```
